```python
competitions_path = 'open-data/data/competitions.json'
with open(competitions_path, 'r') as f:
    competitions = json.load(f)
```

```python
competitions
```

```python
match_to_comp = {}
for comp in competitions:
    comp_id = comp['competition_id']
    season_id = comp['season_id']
    
    matches_path = f'open-data/data/matches/{comp_id}/{season_id}.json'
    try:
        with open(matches_path, 'r') as f:
            matches = json.load(f)
            for match in matches:
                match_to_comp[match['match_id']] = {
                    'competition_id': comp_id,
                    'season_id': season_id,
                    'competition_name': comp['competition_name'],
                    'season_name': comp['season_name'],
                    'country_name': comp['country_name']
                }
    except FileNotFoundError:
        continue
```

```python
# Now collect events properly - extracting match_id from filename
all_events = []
events_dir = 'open-data/data/events'

for filename in os.listdir(events_dir):
    if filename.endswith('.json'):
        match_id = int(filename.replace('.json', ''))  # Extract match_id from filename
        
        with open(os.path.join(events_dir, filename), 'r') as f:
            events = json.load(f)
            
            for event in events:
                # Add match_id to each event
                event['match_id'] = match_id
                
                # Add competition info if available
                if match_id in match_to_comp:
                    event.update(match_to_comp[match_id])
                
                all_events.append(event)

# Create dataframe
events_df = pd.DataFrame(all_events)

# Now process shots or filter as needed
events_df[['match_id', 'player_name', 'competition_name', 'season_name']].head()
```
